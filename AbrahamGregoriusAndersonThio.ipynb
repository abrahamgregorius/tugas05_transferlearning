```python
# Import Library
import random, time, copy, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score, confusion_matrix, ConfusionMatrixDisplay

import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import DataLoader, Dataset
from torchvision import datasets, models, transforms
```

```python
NUM_CLASSES = 10
SEED = 42
PATIENCE = 5
BATCH = 32
EPOCHS = 30


def set_seed(seed):
    random.seed(seed); np.random.seed(seed); torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

def seed_worker(worker_id):
    s = torch.initial_seed() % 2**32
    np.random.seed(s); random.seed(s)

set_seed(SEED)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device, "|", torch.cuda.get_device_name(0) if device.type == "cuda" else "CPU")
if device.type == "cpu":
    warnings.warn("Runs on CPU. Program may run slowly. Change on Runtime -> Change runtime type -> T4 GPU.")
```

```python
raw_train = datasets.OxfordIIITPet(root="./data", split="trainval", download=True)
raw_test  = datasets.OxfordIIITPet(root="./data", split="test", download=True)

CLASS_NAMES = raw_train.classes[:NUM_CLASSES]
all_paths  = list(raw_train._images) + list(raw_test._images)
all_labels = list(raw_train._labels) + list(raw_test._labels)

idx_10 = [i for i, l in enumerate(all_labels) if l < NUM_CLASSES]
paths  = [all_paths[i] for i in idx_10]
labels = [all_labels[i] for i in idx_10]

train_i, temp_i, train_y, temp_y = train_test_split(
    range(len(paths)), labels, test_size=0.30, stratify=labels, random_state=SEED)
val_i, test_i, val_y, test_y = train_test_split(
    temp_i, temp_y, test_size=0.50, stratify=temp_y, random_state=SEED)

assert set(train_i).isdisjoint(val_i) and set(train_i).isdisjoint(test_i) and set(val_i).isdisjoint(test_i)
print(f"Total: {len(paths)} | train: {len(train_i)} | val: {len(val_i)} | test: {len(test_i)}")
print("Class:", CLASS_NAMES)

_pre = transforms.Resize(256)
pil_imgs = [_pre(Image.open(p).convert("RGB")) for p in paths]

class PetDataset(Dataset):
    def __init__(self, indices, transform):
        self.idx = list(indices); self.transform = transform
    def __len__(self): return len(self.idx)
    def __getitem__(self, k):
        i = self.idx[k]
        return self.transform(pil_imgs[i]), labels[i]
```

```python
weights = models.ResNet18_Weights.IMAGENET1K_V1
mean, std = weights.transforms().mean, weights.transforms().std

train_transforms = transforms.Compose([
    transforms.Resize(256),
    transforms.RandomResizedCrop(224, scale=(0.8, 1.0)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(degrees=15),
    transforms.ColorJitter(brightness=0.2, contrast=0.2),
    transforms.ToTensor(),
    transforms.Normalize(mean=mean, std=std),
])

eval_transforms = weights.transforms()
```

```python
STRATEGY = {
    "B1": "Scratch",
    "B2": "Feature extraction ResNet-18",
    "B3": "Progressive FT: layer4 + fc, differential LR",
    "B4_NoAug": "B3 no augmentation",
    "B4_Aug": "B3 with augmentation",
    "B5": "EfficientNet-B0 feature extraction (bonus)",
}

def get_model(code):
    if code == "B1":
        m = models.resnet18(weights=None)
        m.fc = nn.Linear(m.fc.in_features, NUM_CLASSES)
        opt = optim.AdamW([{"params": m.parameters(), "lr": 1e-3, "name": "all"}], weight_decay=1e-2)
    elif code == "B2":
        m = models.resnet18(weights=weights)
        for p in m.parameters(): p.requires_grad = False
        m.fc = nn.Linear(m.fc.in_features, NUM_CLASSES)
        opt = optim.AdamW([{"params": m.fc.parameters(), "lr": 1e-3, "name": "fc"}], weight_decay=1e-2)
    elif code in ("B3", "B4_NoAug", "B4_Aug"):
        m = models.resnet18(weights=weights)
        for p in m.parameters(): p.requires_grad = False
        for p in m.layer4.parameters(): p.requires_grad = True
        m.fc = nn.Linear(m.fc.in_features, NUM_CLASSES)
        opt = optim.AdamW([
            {"params": m.layer4.parameters(), "lr": 1e-4, "name": "layer4"},
            {"params": m.fc.parameters(),     "lr": 1e-3, "name": "fc"},
        ], weight_decay=1e-2)
    elif code == "B5":
        m = models.efficientnet_b0(weights=models.EfficientNet_B0_Weights.IMAGENET1K_V1)
        for p in m.parameters(): p.requires_grad = False
        m.classifier[1] = nn.Linear(m.classifier[1].in_features, NUM_CLASSES)
        opt = optim.AdamW([{"params": m.classifier.parameters(), "lr": 1e-3, "name": "classifier"}], weight_decay=1e-2)
    else:
        raise ValueError(code)
    return m, opt

def set_frozen_bn_eval(model):
    for mod in model.modules():
        if isinstance(mod, nn.modules.batchnorm._BatchNorm) and not any(p.requires_grad for p in mod.parameters()):
            mod.eval()

# Verify jumlah param yang di train
for c in ["B1", "B2", "B3", "B5"]:
    _m, _ = get_model(c)
    tp = sum(p.numel() for p in _m.parameters() if p.requires_grad)
    tt = sum(p.numel() for p in _m.parameters())
    print(f"{c}: trainable {tp:,} / {tt:,}")
    if c == "B2": assert tp == 5130
del _m
```

```python
def make_loader(ds, shuffle, seed):
    g = torch.Generator()
    g.manual_seed(seed)

    return DataLoader(
        ds,
        batch_size=BATCH,
        shuffle=shuffle,
        num_workers=2,
        pin_memory=(device.type == "cuda"),
        generator=g,
        worker_init_fn=seed_worker
    )


def train_one_epoch(model, loader, opt, criterion):
    model.train()
    set_frozen_bn_eval(model)

    tot, correct, n = 0.0, 0, 0

    for x, y in loader:
        x, y = x.to(device, non_blocking=True), y.to(device, non_blocking=True)

        opt.zero_grad(set_to_none=True)

        out = model(x)
        loss = criterion(out, y)

        loss.backward()
        opt.step()

        tot += loss.item() * x.size(0)
        correct += (out.argmax(1) == y).sum().item()
        n += x.size(0)

    return tot / n, correct / n


@torch.no_grad()
def evaluate(model, loader, criterion):
    model.eval()

    tot, probs_all, tgt_all = 0.0, [], []

    for x, y in loader:
        xd, yd = x.to(device, non_blocking=True), y.to(device, non_blocking=True)

        out = model(xd)

        tot += criterion(out, yd).item() * xd.size(0)

        probs_all.append(torch.softmax(out, 1).cpu())
        tgt_all.append(y)

    probs = torch.cat(probs_all).numpy()
    tgt = torch.cat(tgt_all).numpy()
    preds = probs.argmax(1)

    return tot / len(tgt), float((preds == tgt).mean()), preds, tgt, probs


def sync():
    if device.type == "cuda":
        torch.cuda.synchronize()


def run_exp(code, train_tf, seed=SEED):
    set_seed(seed)

    train_loader = make_loader(
        PetDataset(train_i, train_tf), True, seed
    )

    train_eval_loader = make_loader(
        PetDataset(train_i, eval_transforms), False, seed
    )

    val_loader = make_loader(
        PetDataset(val_i, eval_transforms), False, seed
    )

    test_loader = make_loader(
        PetDataset(test_i, eval_transforms), False, seed
    )

    model, opt = get_model(code)
    model.to(device)

    criterion = nn.CrossEntropyLoss()

    trainable_p = sum(
        p.numel() for p in model.parameters() if p.requires_grad
    )

    total_p = sum(p.numel() for p in model.parameters())

    lr_desc = "; ".join(
        f"{g['name']}={g['lr']:g}" for g in opt.param_groups
    )

    print(
        f"\n[{code}] seed={seed} | "
        f"Trainable parameters: {trainable_p:,} / {total_p:,} | "
        f"Learning rates: {lr_desc}"
    )

    history, best_loss, best_epoch, best_state, wait = (
        [], float("inf"), 0, None, 0
    )

    sync()
    t0 = time.time()

    for epoch in range(1, EPOCHS + 1):
        tr_loss, tr_acc = train_one_epoch(
            model, train_loader, opt, criterion
        )

        va_loss, va_acc, *_ = evaluate(
            model, val_loader, criterion
        )

        rec = {
            "Condition": code,
            "Seed": seed,
            "Epoch": epoch,
            "Train_Loss": tr_loss,
            "Train_Acc": tr_acc,
            "Val_Loss": va_loss,
            "Val_Acc": va_acc
        }

        rec.update({
            f"LR_{g['name']}": g["lr"]
            for g in opt.param_groups
        })

        history.append(rec)

        print(
            f"  Epoch {epoch:02d} | "
            f"Train Loss: {tr_loss:.4f} / Train Acc: {tr_acc:.4f} | "
            f"Val Loss: {va_loss:.4f} / Val Acc: {va_acc:.4f}"
        )

        if va_loss < best_loss:
            best_loss, best_epoch, wait = va_loss, epoch, 0
            best_state = copy.deepcopy(model.state_dict())

        else:
            wait += 1

            if wait >= PATIENCE:
                print(
                    f"  Early stopping at epoch {epoch} "
                    f"(best epoch: {best_epoch})"
                )
                break

    sync()
    duration = time.time() - t0
    epochs_run = len(history)

    model.load_state_dict(best_state)

    va_loss_b, va_acc_b, *_ = evaluate(
        model, val_loader, criterion
    )

    _, tr_acc_clean, *_ = evaluate(
        model, train_eval_loader, criterion
    )

    te_loss, te_acc, preds, tgt, probs = evaluate(
        model, test_loader, criterion
    )

    row = {
        "Condition": code,
        "Strategy": STRATEGY[code],
        "Seed": seed,
        "Trainable_Params": trainable_p,
        "Total_Params": total_p,
        "Learning_Rate": lr_desc,
        "Converged_Epoch": best_epoch,
        "Epochs_Run": epochs_run,
        "Duration_Sec": round(duration, 1),
        "Sec_per_Epoch": round(duration / epochs_run, 1),
        "Train_Acc_Clean": round(tr_acc_clean, 4),
        "Val_Acc": round(va_acc_b, 4),
        "Val_Loss": round(va_loss_b, 4),
        "Gen_Gap": round(tr_acc_clean - va_acc_b, 4),
        "Test_Acc": round(te_acc, 4),
        "Test_Macro_F1": round(
            f1_score(tgt, preds, average="macro"), 4
        ),
        "Device": device.type,
    }

    art = {
        "preds": preds,
        "targets": tgt,
        "probs": probs,
        "model": model.to("cpu")
    }

    return row, pd.DataFrame(history), art
```

```python
EXPERIMENTS = [("B1", train_transforms), ("B2", train_transforms), ("B3", train_transforms),
               ("B4_NoAug", eval_transforms), ("B4_Aug", train_transforms), ("B5", train_transforms)]

rows, histories, artifacts = [], {}, {}
for code_, tf in EXPERIMENTS:
    row, hist, art = run_exp(code_, tf)
    rows.append(row); histories[code_] = hist; artifacts[code_] = art

log_df = pd.DataFrame(rows)
log_df.to_csv("experiment_log.csv", index=False)
pd.concat(histories.values(), ignore_index=True).to_csv("history_log.csv", index=False)
print("Saved experiment_log.csv, history_log.csv")
```

```python
FAIL_SELECT = "highest_conf"

N_SHOW = 8

best_code = log_df.loc[log_df["Test_Macro_F1"].idxmax(), "Condition"]

art = artifacts[best_code]

wrong = np.where(art["preds"] != art["targets"])[0]

conf = art["probs"][wrong, art["preds"][wrong]]

print(f"Best model: {best_code} | Number of misclassifications: {len(wrong)} / {len(art['targets'])}")

if FAIL_SELECT == "highest_conf":
    order = wrong[np.argsort(-conf)]

elif FAIL_SELECT == "lowest_conf":
    order = wrong[np.argsort(conf)]

else:
    order = np.random.RandomState(SEED).permutation(wrong)

show = order[:min(N_SHOW, len(order))]

if len(show) < 5:
    warnings.warn("Fewer than 5 misclassified samples; the assignment requires 5–8 samples.")

ncols = 4
nrows = int(np.ceil(len(show) / ncols))

fig, axes = plt.subplots(
    nrows,
    ncols,
    figsize=(4.2 * ncols, 4.4 * nrows)
)
axes = np.atleast_1d(axes).ravel()

rec = []

for ax in axes:
    ax.axis("off")

for ax, k in zip(axes, show):

    t, p = int(art["targets"][k]), int(art["preds"][k])
    cf = float(art["probs"][k, p])

    ax.imshow(pil_imgs[test_i[k]])

    ax.set_title(
        f"True: {CLASS_NAMES[t]}\nPredicted: {CLASS_NAMES[p]} ({cf:.2f})",
        color="red",
        fontsize=10
    )

    rec.append({
        "Test_Idx": int(k),
        "Image_File": paths[test_i[k]].name,
        "True": CLASS_NAMES[t],
        "Pred": CLASS_NAMES[p],
        "Confidence": round(cf, 4),
        "Prob_True_Class": round(float(art["probs"][k, t]), 4),
        "Hypothesis": ""
    })

plt.suptitle(
    f"Test Set Misclassifications — Model {best_code}",
    y=1.0
)

plt.tight_layout()
plt.savefig("failure_analysis.png", dpi=150)
plt.show()

fail_df = pd.DataFrame(rec)
fail_df.to_csv("failure_analysis.csv", index=False)
display(fail_df)
```

```python
def gradcam(model, x, target):

    model.eval()
    store = {}

    def fh(m, i, o):
        o.retain_grad()
        store["a"] = o

    h = model.layer4.register_forward_hook(fh)

    x = x.clone().requires_grad_(True)

    with torch.enable_grad():

        out = model(x)
        model.zero_grad()
        out[0, target].backward()

    h.remove()

    a, g = store["a"], store["a"].grad

    cam = torch.relu(
        (g.mean(dim=(2, 3), keepdim=True) * a).sum(1, keepdim=True)
    )

    cam = F.interpolate(
        cam,
        size=(224, 224),
        mode="bilinear",
        align_corners=False
    )[0, 0].detach().cpu().numpy()

    return (cam - cam.min()) / (cam.max() - cam.min() + 1e-8)


m_before, m_after = artifacts["B2"]["model"].to(device), artifacts["B3"]["model"].to(device)

correct_idx = np.where(art["preds"] == art["targets"])[0]

pick = list(show[:3]) + list(correct_idx[:3])

mean_t, std_t = torch.tensor(mean).view(3, 1, 1), torch.tensor(std).view(3, 1, 1)

fig, axes = plt.subplots(len(pick), 3, figsize=(9, 3 * len(pick)))

for r, k in enumerate(pick):

    xk = eval_transforms(pil_imgs[test_i[k]]).unsqueeze(0).to(device)
    t = int(art["targets"][k])

    img = (
        (xk[0].cpu() * std_t + mean_t)
        .clamp(0, 1)
        .permute(1, 2, 0)
        .numpy()
    )

    axes[r, 0].imshow(img)
    axes[r, 0].set_title(f"True: {CLASS_NAMES[t]}", fontsize=9)

    for c, (mdl, name) in enumerate([
        (m_before, "B2 (Before Fine-Tuning)"),
        (m_after, "B3 (After Fine-Tuning)")
    ], start=1):

        axes[r, c].imshow(img)
        axes[r, c].imshow(gradcam(mdl, xk, t), cmap="jet", alpha=0.45)
        axes[r, c].set_title(name, fontsize=9)

    for c in range(3):
        axes[r, c].axis("off")

plt.tight_layout()
plt.savefig("gradcam_before_after.png", dpi=150)
plt.show()

artifacts["B2"]["model"].to("cpu")
artifacts["B3"]["model"].to("cpu")
```
